# Langkah 1: Import Library & Load Dataset
Pertama kita import library yang diperlukan dan membaca dataset.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from IPython.display import display  # supaya tabel lebih rapi

# Load dataset
df = pd.read_csv(r"C:\Users\ACER\OneDrive\Documents\semester5\kb\archive (1)\student_performance.csv")

# Atur opsi tampilan pandas
pd.set_option("display.max_columns", None)   # tampilkan semua kolom
pd.set_option("display.width", 1000)         # biar tidak terpotong
pd.set_option("display.colheader_justify", "center")  # header rata tengah

print("Data awal:")
display(df.head())  # lebih rapi daripada print
```

# Langkah 2: Data Cleaning
Pada tahap ini dilakukan:
- Handling Missing Value → mengisi nilai kosong
- Handling Duplicate Value → menghapus data ganda
- Handling Outlier → membuang data yang terlalu jauh dari distribusi normal

```python
from IPython.display import display

# Cek missing value
print("Jumlah missing value per kolom:")
display(df.isnull().sum().to_frame("Missing Values"))

# Isi nilai kosong numerik dengan mean, kategorikal dengan modus
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].fillna(df[col].mode()[0])
    else:
        df[col] = df[col].fillna(df[col].mean())

# Hapus duplicate jika ada
df = df.drop_duplicates()

# Handling outlier dengan metode IQR
for col in df.select_dtypes(include=['int64', 'float64']).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower, upper)

print("\nData setelah cleaning:")
display(df.head())
```

# Langkah 3: Normalisasi / Standarisasi
Standarisasi digunakan untuk mengubah skala data numerik menjadi distribusi dengan mean = 0 dan standar deviasi = 1.

```python
from IPython.display import display
from sklearn.preprocessing import StandardScaler

# Standarisasi kolom numerik
scaler = StandardScaler()
num_cols = df.select_dtypes(include=np.number).columns
df[num_cols] = scaler.fit_transform(df[num_cols])

print("Data setelah standarisasi (contoh 5 baris):")
display(df.head())
```

# Langkah 4: Encoding Kolom Kategorikal
Kolom kategorikal (grade) diubah menjadi angka menggunakan Label Encoding.

```python
from sklearn.preprocessing import LabelEncoder
from IPython.display import display

encoder = LabelEncoder()
if 'grade' in df.columns:
    df['grade_encoded'] = encoder.fit_transform(df['grade'])
    print("Contoh hasil encoding kolom grade:")
    display(df[['grade', 'grade_encoded']].head())
```

#  A1. Linear Regression dan Support Vector Regressor (SVR)

Pada bagian ini dilakukan **analisis regresi numerik** untuk memprediksi nilai `total_score` siswa berdasarkan fitur:
- `weekly_self_study_hours`
- `attendance_percentage`
- `class_participation`
- `performance_index`

Model yang digunakan:
1. **Linear Regression**
2. **Support Vector Regressor (SVR)** sebagai pembanding.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Gunakan hanya 200 baris pertama agar proses cepat
df_subset = df.head(200)

# Fitur dan target
X_reg = df_subset[['weekly_self_study_hours', 'attendance_percentage', 'class_participation', 'performance_index']]
y_reg = df_subset['total_score']

# Split data 80:20
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

print("Jumlah data latih:", X_train_reg.shape[0])
print("Jumlah data uji  :", X_test_reg.shape[0])
X_train_reg.head()
```

## Pelatihan Model Regresi
Kita akan melatih dua model:
1. **Linear Regression**
2. **Support Vector Regressor (SVR)** dengan kernel linear


```python
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR

# Linear Regression
lin_reg = LinearRegression()
lin_reg.fit(X_train_reg, y_train_reg)
y_pred_lin = lin_reg.predict(X_test_reg)

# SVR
svr = SVR(kernel='linear')
svr.fit(X_train_reg, y_train_reg)
y_pred_svr = svr.predict(X_test_reg)

# Tampilkan contoh hasil prediksi
hasil_df = pd.DataFrame({
    "Actual": y_test_reg.values[:5],
    "Predicted (Linear)": y_pred_lin[:5],
    "Predicted (SVR)": y_pred_svr[:5]
})
hasil_df
```

#  A2. Evaluasi Model Regresi

## Evaluasi Kinerja Model
menghitung metrik evaluasi untuk kedua model:
- **MAE** (Mean Absolute Error)
- **MSE** (Mean Squared Error)
- **RMSE** (Root Mean Squared Error)
- **R²** (Koefisien Determinasi)


Semakin kecil nilai error dan semakin besar R², semakin baik performa model.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

def evaluasi_model(y_test, y_pred, nama_model):
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    return pd.DataFrame({
        "Model": [nama_model],
        "MAE": [mae],
        "MSE": [mse],
        "RMSE": [rmse],
        "R²": [r2]
    })

# Gabungkan hasil evaluasi dua model
hasil_eval = pd.concat([
    evaluasi_model(y_test_reg, y_pred_lin, "Linear Regression"),
    evaluasi_model(y_test_reg, y_pred_svr, "Support Vector Regressor (SVR)")
], ignore_index=True)

print(" Hasil Evaluasi Model:")
display(hasil_eval)
```

## Analisis Hasil Evaluasi
Interpretasi:

Linear Regression memberikan hasil lebih baik (R² lebih tinggi, error lebih kecil).

SVR sedikit kalah karena kernel linear tidak cukup menangkap variasi data yang mungkin masih kompleks.

Kedua model sama-sama cukup stabil untuk dataset kecil (200 baris).

# A3. Visualisasi Hasil Prediksi
## Visualisasi Hasil Prediksi
Visualisasi ini memperlihatkan hubungan antara **nilai aktual** dan **prediksi** untuk kedua model.
Garis putus-putus hitam menunjukkan prediksi sempurna.


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(y_test_reg, y_pred_lin, color='blue', alpha=0.7, label='Linear Regression')
plt.scatter(y_test_reg, y_pred_svr, color='red', alpha=0.6, label='SVR')
plt.plot([y_test_reg.min(), y_test_reg.max()], [y_test_reg.min(), y_test_reg.max()], 'k--', lw=2)
plt.title("Perbandingan Nilai Aktual vs Prediksi")
plt.xlabel("Nilai Aktual Total Score")
plt.ylabel("Nilai Prediksi Total Score")
plt.legend()
plt.grid(True)
plt.show()
```

Scatter plot menunjukkan titik-titik prediksi dekat dengan garis diagonal, artinya prediksi cukup akurat.

Namun, ada beberapa outlier (titik jauh dari garis), menunjukkan ada sebagian data yang sulit diprediksi dengan baik oleh kedua model.

Secara umum, Linear Regression (biru) lebih mendekati garis dibanding SVR (merah).
